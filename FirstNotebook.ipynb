## Filtering the best player

```python
import pandas as pd

df = pd.read_csv("IPL_2026_GroupStage_Stats.csv", index_col="POS")

# 1. Set display options to show ALL rows and columns
pd.set_option('display.max_rows', None)      # None means unlimited rows
pd.set_option('display.max_columns', None)   # None means unlimited columns
pd.set_option('display.width', 1000)         # Prevents columns from wrapping to a new line

desired_runs = int(input("Enter Min Runs: "))
desired_sr = int(input("Enter Min SR: "))
desired_avg = int(input("Enter Min Avg: "))

criteria = df[(df["Runs"]>= desired_runs)&
              (df["SR"]>= desired_sr) &
              (df["Avg"]>= desired_avg)]

print(criteria)
```

## Best Finisher

```python
#Program to find the best finisher in IPL 2026
import pandas as pd

df = pd.read_csv("IPL_2026_GroupStage_Stats.csv", index_col="POS")

# 1. Set display options to show ALL rows and columns
pd.set_option('display.max_rows', None)      # None means unlimited rows
pd.set_option('display.max_columns', None)   # None means unlimited columns
pd.set_option('display.width', 1000)         # Prevents columns from wrapping to a new line

inns_to_no_ratio = df["Inns"] / df["NO"]

runs_to_inns_ratio = df["Runs"] / df["Inns"] 

ratio = (inns_to_no_ratio <= (1/3)*14) & (runs_to_inns_ratio >= 20 ) & (df["SR"]>=165)

filtered_df = df[ratio]

final_result = filtered_df.sort_values(by='SR', ascending=False)

print("BEST FINISHERS OF IPL 2026")
print(final_result)

```
