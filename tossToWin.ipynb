# Tosses Won Leading to Wins

### Row by Row Approach

```python
import pandas as pd

df = pd.read_csv("Match_Info.csv")

df_filtered = df.iloc[:10, :9]

print(df_filtered)

num = 0
total_matches = 0

#for loop goes row by row
#iterrows() returns the index (i) and the row data (row) for each row
for i, row in df_filtered.iterrows():
    if row["toss_winner"] == row["winner"]:
        num +=1
        total_matches+=1
    else: 
        total_matches+=1
print(f"\nTotal matches won by toss winner: {num} out of {total_matches} matches")    

```

## Vectorization

```python
import pandas as pd

df = pd.read_csv("Match_Info.csv")

# 1. Compare the two columns directly
# This creates a Series of True/False values for every row
matches = df["toss_winner"] == df["winner"]

# 2. Sum the True values (True = 1, False = 0)
wins = matches.sum()

total_matches = len(df)

win_percentage = (wins/ total_matches)*100

print(f"Total toss-winners won {wins} out of {total_matches} matches; equating to a win percentage of {win_percentage}%")
```

## Field vs Bat

```python
import pandas as pd

df = pd.read_csv("Match_Info.csv")

# 1. Compare the two columns directly
# This creates a Series of True/False values for every row
matches_bat = (df["toss_winner"] == df["winner"]) & (df['toss_decision'] == "bat")

# 2. Sum the True values (True = 1, False = 0)
wins_bat = matches_bat.sum()

total_matches_bat = (df['toss_decision'] == "bat").sum()

win_percentage_bat = (wins_bat/ total_matches_bat)*100

print(f"Total toss-winners that chose batting won {wins_bat} out of {total_matches_bat} matches; equating to a win percentage of {win_percentage_bat}%")

matches_field = (df["toss_winner"] == df["winner"]) & (df['toss_decision'] == "field")

# 2. Sum the True values (True = 1, False = 0)
wins_field = matches_field.sum()

total_matches_field = (df['toss_decision'] == "field").sum()

win_percentage_field = (wins_field/ total_matches_field)*100

print(f"Total toss-winners that chose fielding won {wins_field} out of {total_matches_field} matches; equating to a win percentage of {win_percentage_field}%")
```

## RCB Specific

```python
import pandas as pd

df = pd.read_csv("Match_Info.csv")

matches = (df["toss_winner"] == "Royal Challengers Bangalore") & \
          (df["winner"] == "Royal Challengers Bangalore") & \
          ((df['team1'] == "Royal Challengers Bangalore") | (df['team2'] == "Royal Challengers Bangalore" ))

wins = matches.sum()

total_matches = ((df['team1'] == "Royal Challengers Bangalore") | (df['team2'] == "Royal Challengers Bangalore" )).sum()

win_percentage = (wins/total_matches)*100

print(f"RCB won {wins} matches out of {total_matches} when they won the toss; this equates to a win percentage of {win_percentage}%")
```
