# Tosses Won Leading to Wins

### Row by Row Approach

```python
import pandas as pd

df = pd.read_csv("Match_Info.csv")

df_filtered = df.iloc[:10, :9]

print(df_filtered)

num = 0
total_matches = 0

#for loop goes row by row
#iterrows() returns the index (i) and the row data (row) for each row
for i, row in df_filtered.iterrows():
    if row["toss_winner"] == row["winner"]:
        num +=1
        total_matches+=1
    else: 
        total_matches+=1
print(f"\nTotal matches won by toss winner: {num} out of {total_matches} matches")    

```

## Vectorization

```python
import pandas as pd

df = pd.read_csv("Match_Info.csv")

# 1. Compare the two columns directly
# This creates a Series of True/False values for every row
matches = df["toss_winner"] == df["winner"]

# 2. Sum the True values (True = 1, False = 0)
wins = matches.sum()

total_matches = len(df)

win_percentage = (wins/ total_matches)*100

print(f"Total toss-winners won {wins} out of {total_matches} matches; equating to a win percentage of {win_percentage}%")
```
